```python
import statistics

# 1 ile 100 arasındaki çift sayıları liste olarak oluşturuyoruz
# 2'den başlar, 101'e kadar gider (100 dahil), 2'şer artar
sayilar = list(range(2, 101, 2))

# Hesaplamalar
aritmetik_ort = statistics.mean(sayilar)
geometrik_ort  = statistics.geometric_mean(sayilar)
harmonik_ort   = statistics.harmonic_mean(sayilar)
medyan         = statistics.median(sayilar)
standart_sapma = statistics.stdev(sayilar)
varyans        = statistics.variance(sayilar)

# Mod hesaplama (Çift sayılar dizisinde her sayıdan birer tane olduğu için mod tüm dizidir)
# statistics.multimode() tüm sayıları döndürecektir.
mod = statistics.multimode(sayilar)

# Sonuçları Yazdırma
print(f"Sayı Adedi: {len(sayilar)}")
print("-" * 30)
print(f"Aritmetik Ortalama: {aritmetik_ort}")
print(f"Geometrik Ortalama: {geometrik_ort:.2f}")
print(f"Harmonik Ortalama:  {harmonik_ort:.2f}")
print(f"Medyan (Ortanca):   {medyan}")
print(f"Standart Sapma:     {standart_sapma:.2f}")
print(f"Varyans:            {varyans:.2f}")
print(f"Mod:                {mod if len(mod) < len(sayilar) else 'Her sayıdan birer adet var.'}")
```

**VERİ SETİ OLUŞTURMA**

np.random.normal, Python'un NumPy kütüphanesi içinde yer alan ve Normal Dağılım (diğer adıyla Çan Eğrisi veya Gaussian Dağılımı) kurallarına göre rastgele sayılar üreten bir fonksiyondur.

Doğadaki pek çok olgu (insanların boy uzunluğu, sınav notları, ölçüm hataları) tamamen rastgele değil, bir ortalamanın etrafında kümelenmiş şekildedir. Bu fonksiyon, bu doğal dağılımı simüle etmemizi sağlar.

**Fonksiyonun Yapısı ve Parametreleri**

Genellikle şu şekilde kullanılır:

`np.random.normal(loc,scale,size)`

**loc (Mean - Ortalama):** Dağılımın merkezidir. Çan eğrisinin en tepe noktasının nerede olacağını belirler. Örneğin 65 yazarsanız, sayıların çoğu 65 civarında oluşur.

**scale (Standard Deviation - Standart Sapma):** Dağılımın genişliğini/yayılımını belirler.

Bu sayı küçükse, üretilen sayılar ortalamaya çok yakındır (dik bir çan).

Bu sayı büyükse, sayılar ortalamadan uzağa yayılır (basık bir çan).

**size (Gözlem Sayısı):** Toplamda kaç adet rastgele sayı üretileceğini belirler (Örn: 100 öğrenci).

image.png

`np.random.seed(42)` , bilgisayarın rastgele sayı üretme sürecini sabit bir başlangıç noktasına bağlayan komuttur. Bilgisayarlar aslında tam olarak "rastgele" sayı üretemezler; bunun yerine belirli bir algoritma (formül) üzerinden karmaşık sayılar dizisi oluştururlar. seed (tohum) ise bu formülün başlangıç değeridir.

İşte bu komutun neden ve nasıl kullanıldığına dair detaylar:

**1. Neden Kullanılır? (Tekrar Edilebilirlik)**

Bir veri seti oluşturduğunuzda veya bir model eğittiğinizde, kodunuzu her çalıştırdığınızda farklı sayılar gelirse sonuçlarınız sürekli değişir.

Seed kullanmazsanız: Birinci çalıştırmada vize ortalaması 64, ikinci çalıştırmada 66 çıkar.

Seed kullanırsanız: Kodunuzu ister bugün ister 3 yıl sonra çalıştırın, her zaman aynı rastgele sayı dizisini elde edersiniz.

Bu durum, akademik çalışmalarda ve hata ayıklama (debugging) süreçlerinde sonuçların tutarlı ve doğrulanabilir olması için kritiktir.

**2. "42" Sayısının Bir Gizemi Var mı?**

Aslında hayır. Oraya 1, 100, 1903 veya herhangi bir tam sayı yazabilirsiniz.

Önemli olan sayı değil, o sayının sabit kalmasıdır.

42 sayısı, popüler kültürde (Otostopçunun Galaksi Rehberi kitabına atıfla) "hayatın, evrenin ve her şeyin cevabı" olarak kabul edildiği için yazılımcılar arasında bir gelenek haline gelmiştir.

**3. Nasıl Çalışır?**

Bunu bir kitap kütüphanesi gibi düşünebilirsiniz:

`np.random.seed(42)` dediğinizde, kütüphanedeki 42 numaralı rafı seçmiş olursunuz.

Ardından `np.random.normal()` dediğinizde, o raftaki kitapları sırayla okumaya başlarsınız.

Programı kapatıp tekrar açıp yine seed(42) derseniz, bilgisayar yine 42 numaralı rafa gider ve aynı kitapları (sayıları) aynı sırayla okur.

```python
import pandas as pd
import numpy as np

# Rastgelelik için sabit bir seed belirliyoruz
np.random.seed(42)

# 100 öğrenci için ID listesi
ogrenci_id = np.arange(1001, 1101)

# Notları normal dağılıma göre oluşturalım (Ortalama 65, Standart Sapma 15)
vize_notlari = np.random.normal(65, 15, 100).clip(0, 100).astype(int)
odev_notlari = np.random.normal(75, 10, 100).clip(0, 100).astype(int)
final_notlari = np.random.normal(60, 20, 100).clip(0, 100).astype(int)

# DataFrame oluşturma
df = pd.DataFrame({
    'Ogrenci_ID': ogrenci_id,
    'Vize': vize_notlari,
    'Odev': odev_notlari,
    'Final': final_notlari
})

# Başarı notunu hesaplama (%30 Vize, %20 Ödev, %50 Final)
df['Basari_Notu'] = (df['Vize'] * 0.3 + df['Odev'] * 0.2 + df['Final'] * 0.5).round(2)

# Veri setinin ilk 5 satırı
print(df.head())
```

**1101 (Bitiş):** Dizinin duracağı sınır değerdir. Bu sayı diziye dahil değildir. (Yani en son 1100 üretilir).

**Artış Miktarı (Varsayılan):** Üçüncü bir parametre yazılmadığı sürece sayılar 1'er 1'er artar.

**Sonuç Ne Olur?**

Bu komutu çalıştırdığımızda bilgisayar bize şu listeyi üretir:
[1001, 1002, 1003, ..., 1100]

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats

# -------------------
# Veri Yükleme
# -------------------
df = pd.read_csv('ogrenci_notlari.csv')
df.columns = df.columns.str.strip()

notlar = df["Basari_Notu"].dropna()

# -------------------
# Tanımlayıcı İstatistikler
# -------------------
n = len(notlar)
mean = np.mean(notlar)
median = np.median(notlar)
std = np.std(notlar, ddof=1)
var = np.var(notlar, ddof=1)
min_val = np.min(notlar)
max_val = np.max(notlar)

skewness = stats.skew(notlar)
kurtosis = stats.kurtosis(notlar)

# -------------------
# Güven Aralığı (%95)
# -------------------
confidence = 0.95
t_crit = stats.t.ppf((1 + confidence) / 2, df=n-1)
margin_error = t_crit * std / np.sqrt(n)
ci_lower = mean - margin_error
ci_upper = mean + margin_error

# -------------------
# Normallik Testleri
# -------------------
shapiro_stat, shapiro_p = stats.shapiro(notlar)
jb_stat, jb_p = stats.jarque_bera(notlar)

# -------------------
# Aykırı Değer (IQR)
# -------------------
Q1 = np.percentile(notlar, 25)
Q3 = np.percentile(notlar, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = notlar[(notlar < lower_bound) | (notlar > upper_bound)]

# -------------------
# Z-score ile aykırı değer
# -------------------
z_scores = np.abs(stats.zscore(notlar))
outliers_z = notlar[z_scores > 3]

# -------------------
# SONUÇLARI YAZDIR
# -------------------
print("=== TANIMLAYICI İSTATİSTİKLER ===")
print(f"Gözlem sayısı: {n}")
print(f"Ortalama: {mean:.2f}")
print(f"Medyan: {median:.2f}")
print(f"Standart sapma: {std:.2f}")
print(f"Varyans: {var:.2f}")
print(f"Min-Max: {min_val} - {max_val}")

print("\n=== DAĞILIM ===")
print(f"Skewness: {skewness:.3f}")
print(f"Kurtosis: {kurtosis:.3f}")

print("\n=== %95 GÜVEN ARALIĞI ===")
print(f"[{ci_lower:.2f}, {ci_upper:.2f}]")

print("\n=== NORMALİK TESTLERİ ===")
print(f"Shapiro p: {shapiro_p:.4f}")
print(f"Jarque-Bera p: {jb_p:.4f}")

print("\n=== AYKIRI DEĞERLER ===")
print(f"IQR yöntemi: {len(outliers_iqr)} adet")
print(f"Z-score yöntemi: {len(outliers_z)} adet")

# -------------------
# GRAFİKLER
# -------------------
fig, axes = plt.subplots(3, 1, figsize=(8, 14))

# Histogram + KDE
axes[0].hist(notlar, bins=10, density=True)
notlar.plot(kind='kde', ax=axes[0])
axes[0].axvline(mean, linestyle='--', label=f"Ortalama: {mean:.2f}")
axes[0].axvline(median, linestyle='-', label=f"Medyan: {median:.2f}")
axes[0].set_title("Histogram + KDE")
axes[0].legend()

# Boxplot
axes[1].boxplot(notlar, vert=True)
axes[1].set_title("Boxplot (Aykırı Değer Analizi)")

# Q-Q Plot
stats.probplot(notlar, dist="norm", plot=axes[2])
axes[2].set_title("Q-Q Plot")

plt.tight_layout()
plt.show()

# -------------------
# OTOMATİK AKADEMİK YORUM
# -------------------
print("\n=== AKADEMİK YORUM ===")

# Normallik
if shapiro_p > 0.05:
    print("Veri normal dağılıma uygundur (Shapiro-Wilk).")
else:
    print("Veri normal dağılıma uygun değildir (Shapiro-Wilk).")

# Çarpıklık
if skewness > 0:
    print("Dağılım sağa çarpıktır.")
elif skewness < 0:
    print("Dağılım sola çarpıktır.")
else:
    print("Dağılım simetriktir.")

# Basıklık
if kurtosis > 3:
    print("Dağılım leptokurtiktir (sivri).")
else:
    print("Dağılım platykurtiktir (basık).")

# Aykırı değer
if len(outliers_iqr) > 0:
    print("Veride aykırı değerler bulunmaktadır.")
else:
    print("Belirgin aykırı değer yoktur.")
```

**1. Temel Çoklu Doğrusal Regresyon (OLS)**

```python
import pandas as pd
import statsmodels.api as sm

# Veri
df = pd.read_csv('ogrenci_notlari.csv')
df.columns = df.columns.str.strip()

# Değişkenler
X = df[["Vize", "Odev", "Final"]]
y = df["Basari_Notu"]

# Sabit terim ekle
X = sm.add_constant(X)

# Model
model = sm.OLS(y, X).fit()

# Sonuç
print(model.summary())
```

$Basari\_Notu = \beta_0 + \beta_1 \cdot Vize + \beta_2 \cdot Odev + \beta_3 \cdot Final$



$\beta \text{ katsayıları} \rightarrow \text{hangi değişken daha etkili}$


P-Değeri ve Anlamlılık:
p-value<0.05→istatistiksel olarak anlamlı

$Belirleme\qquad Katsayısı (R^2)$

$R^2 \approx 0.80 \implies \text{Model verinin } \%80\text{'ini açıklıyor (Çok İyi)}$

$0.70 < R^2 < 1.00   \qquad  \text {(Çok İyi / Güçlü İlişki)}$


$0.40<R2<0.70 \qquad \text{Orta Düzey İlişki}$

$R2<0.40  \qquad\text{Zayıf İlişki / Düşük Açıklayıcılık}$



**2. Varsayım Kontrolleri (AKADEMİK OLARAK ŞART)**

 Artıklar (residuals)

```python
import matplotlib.pyplot as plt

residuals = model.resid

plt.hist(residuals, bins=10)
plt.title("Artıkların Dağılımı")
plt.show()
```

**Q-Q plot (normallik)**

```python
import scipy.stats as stats

stats.probplot(residuals, dist="norm", plot=plt)
plt.title("Artıklar Q-Q Plot")
plt.show()
```

**Heteroskedastisite (Breusch-Pagan)**

```python
from statsmodels.stats.diagnostic import het_breuschpagan

test = het_breuschpagan(residuals, X)
print("p-değeri:", test[1])
```

p < 0.05 → varyans sabit değil (problem var)

**3. Çoklu Bağlantı (Multicollinearity - VIF)**

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_data = pd.DataFrame()
vif_data["Değişken"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

print(vif_data)
```

VIF > 10 → ciddi problem

Çoklu doğrusal regresyon analizi sonucunda modelin anlamlı olduğu (p < 0.05) görülmüştür. Modelin açıklayıcılık gücü R² değeri ile değerlendirilmiş ve bağımsız değişkenlerin bağımlı değişken üzerindeki etkileri katsayılar aracılığıyla yorumlanmıştır. Ayrıca model varsayımları (normallik, homoskedastisite ve çoklu bağlantı) test edilmiştir.

1. Regresyon Denklemini Grafikle Gösterme (Regression Plot)

Bağımsız değişkenler ile bağımlı değişken arasındaki ilişkiyi sadece sayılarla değil, regresyon doğrusuyla görmek modelin başarısını (R²) görsel olarak kanıtlar. seaborn kütüphanesi bu iş için biçilmiş kaftandır.

2. Standartlaştırılmış Katsayılar (β)

Ham katsayılar birimlere (vize puanı, ödev puanı vb.) bağlıdır. Değişkenlerin etki gücünü gerçek anlamda kıyaslamak için veriyi standardize edip (μ=0,σ=1) katsayıları öyle hesaplarız. Böylece "Vize mi daha etkili yoksa Ödev mi?" sorusuna bilimsel bir cevap veririz.

3. Model Karşılaştırma (AIC, BIC)

Farklı değişken kombinasyonlarını (örneğin sadece Vize-Final vs Vize-Ödev-Final) test ederken AIC (Akaike Information Criterion) ve BIC (Bayesian Information Criterion) değerlerini kullanırız.

Kural: AIC ve BIC değeri ne kadar düşükse, model o kadar verimli ve dengelidir.

4. Makine Öğrenmesi ile Karşılaştırma

İstatistiksel regresyon (Ordinary Least Squares) ile Makine Öğrenmesi (Scikit-learn LinearRegression) yaklaşımlarını karşılaştırıp, modelin eğitim ve test verisi üzerindeki hata paylarını (MSE, MAE) hesaplarız.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

# Veriyi yüklüyoruz
df = pd.read_csv('ogrenci_notlari.csv')
X = df[['Vize', 'Odev', 'Final']]
y = df['Basari_Notu']

# --- A. İSTATİSTİKSEL MODEL (statsmodels) ---
X_with_const = sm.add_constant(X) # Sabit terim (beta0) ekleme
model_stats = sm.OLS(y, X_with_const).fit()

print(f"AIC: {model_stats.aic:.2f}, BIC: {model_stats.bic:.2f}")

# --- B. STANDARTLAŞTIRILMIŞ KATSAYILAR (Beta) ---
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
model_beta = sm.OLS(y, X_scaled).fit() # Standardize edilmiş katsayılar

# --- C. GÖRSELLEŞTİRME (Regresyon Grafiği) ---
plt.figure(figsize=(8, 5))
sns.regplot(x='Vize', y='Basari_Notu', data=df, scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Vize Notu vs Başarı Notu Regresyon Analizi')
plt.savefig('regresyon_grafigi.png')

# --- D. MAKİNE ÖĞRENMESİ (sklearn) ---
ml_model = LinearRegression()
ml_model.fit(X, y)
print(f"ML Model R^2 Skoru: {ml_model.score(X, y):.4f}")
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

# 1. Veriyi Hazırlama
df = pd.read_csv('ogrenci_notlari.csv')
X = df[['Vize', 'Odev', 'Final']]
y = df['Basari_Notu']

# --- A. REGRESYON GRAFİĞİ (Görselleştirme) ---
plt.figure(figsize=(10, 6))
# Başarı notu ile en güçlü korelasyona sahip olması beklenen 'Final' arasındaki ilişki
sns.regplot(x='Final', y='Basari_Notu', data=df,
            scatter_kws={'color': '#000080', 'alpha': 0.5},
            line_kws={'color': '#F44336'})
plt.title('Final Notu ve Başarı Notu Regresyon Doğrusu')
plt.savefig('regresyon_analizi.png')
print("1. Regresyon grafiği kaydedildi.")

# --- B. İSTATİSTİKSEL MODEL (AIC, BIC ve P-Değerleri) ---
X_stat = sm.add_constant(X) # Sabit terim beta0 ekleniyor
model_ols = sm.OLS(y, X_stat).fit()

print("\n--- Model Performans Kriterleri ---")
print(f"AIC Değeri: {model_ols.aic:.2f}")
print(f"BIC Değeri: {model_ols.bic:.2f}")
print(f"R-Kare (R2): {model_ols.rsquared:.4f}")

# --- C. STANDARTLAŞTIRILMIŞ KATSAYILAR (Beta) ---
# Değişkenlerin etki gücünü birimden bağımsız kıyaslamak için
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
model_beta = sm.OLS(y, X_scaled).fit()

print("\n--- Standartlaştırılmış Katsayılar (Beta) ---")
# Bu katsayılar hangi değişkenin başarı notunu daha çok 'itelediğini' gösterir
for i, col in enumerate(X.columns):
    print(f"{col} Etki Katsayısı: {model_beta.params[i]:.4f}")

# --- D. MAKİNE ÖĞRENMESİ KARŞILAŞTIRMASI ---
ml_reg = LinearRegression()
ml_reg.fit(X, y)
y_pred = ml_reg.predict(X)

print("\n--- Makine Öğrenmesi (LinearRegression) vs OLS ---")
print(f"ML MSE (Hata Kareler Ortalaması): {mean_squared_error(y, y_pred):.4f}")
print(f"ML R2 Skoru: {r2_score(y, y_pred):.4f}")
```
